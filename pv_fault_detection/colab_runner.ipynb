```python
# CELL 6: Build data index (downloads IRSM if not already in zip)
import subprocess, sys, os
import pandas as pd

COLAB_PROJECT = '/content/pv_fault_detection'
index_file = 'data/processed/index/full.parquet'

if os.path.exists(index_file):
    df = pd.read_parquet(index_file)
    print(f'Index already present: {len(df):,} samples. Skipping download.')
    print(f'Domains: {df["domain_id"].value_counts().to_dict()}')
else:
    print('Index not found. Downloading IRSM dataset (~2.3 GB, takes 10-15 min)...')

    # Try project downloader first
    r = subprocess.run(
        [sys.executable, '-m', 'src.data.download', '--dataset', 'irsm'],
        capture_output=False, text=True
    )

    if r.returncode != 0 or not os.path.exists('data/raw/irsm'):
        print('Downloader script unavailable. Downloading directly from GitHub Releases...')
        import urllib.request, zipfile
        os.makedirs('data/raw/irsm', exist_ok=True)
        URL = ('https://github.com/RaptorMaps/InfraredSolarModules/releases/'
               'download/v0.0.1/InfraredSolarModules.zip')
        print(f'URL: {URL}')
        def show_progress(count, block, total):
            if total > 0:
                pct = min(count * block / total * 100, 100)
                print(f'\r  {pct:.1f}%  ({count * block / 1e6:.0f}/{total / 1e6:.0f} MB)',
                      end='', flush=True)
        urllib.request.urlretrieve(URL, '/tmp/irsm.zip', reporthook=show_progress)
        print()
        print('Extracting...')
        with zipfile.ZipFile('/tmp/irsm.zip', 'r') as z:
            z.extractall('data/raw/irsm')
        os.remove('/tmp/irsm.zip')

    # Build index
    print('Building canonical index...')
    r2 = subprocess.run(
        [sys.executable, '-m', 'src.data.indexer', '--config', 'configs/base.yaml'],
        capture_output=True, text=True
    )
    print(r2.stdout[-2000:])
    if r2.returncode != 0:
        print('ERROR:', r2.stderr[-500:])
        raise RuntimeError('Indexer failed - check output above')

    df = pd.read_parquet(index_file)
    print(f'\nIndex built: {len(df):,} samples')

print(open('data/DATASET_STATUS.md').read())
print('Data ready!')
```

```python
# CELL 9: Experiment runner helper (MUST RUN before any experiment cells)
import subprocess, sys, json
from pathlib import Path

def run_experiment(config: str, extra_overrides=None, label='') -> dict:
    """Run one experiment, stream output, return metrics dict."""
    overrides = GPU_OVERRIDES.copy()
    if extra_overrides:
        overrides.extend(extra_overrides)
    cmd = [sys.executable, '-m', 'src.experiments.run',
           '--config', config, '--override'] + overrides
    tag = label or Path(config).stem
    print(f'\n{'='*55}\n{tag}\n{'='*55}')
    proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                            text=True, bufsize=1)
    for line in proc.stdout:
        print(line, end='', flush=True)
    proc.wait()
    if proc.returncode != 0:
        print(f'[exit code {proc.returncode}]')
        return {}
    # Load metrics from most-recent result dir for this config
    cfg_name = Path(config).stem
    dirs = sorted([d for d in Path('results').iterdir()
                   if d.is_dir() and cfg_name in d.name], reverse=True)
    if dirs and (dirs[0] / 'metrics.json').exists():
        m = json.load(open(dirs[0] / 'metrics.json'))
        f1 = m.get('macro_f1') or m.get('test_metrics', {}).get('macro_f1', 'N/A')
        print(f'\nRESULT: Macro-F1 = {f1}')
        return m
    return {}

print('run_experiment() helper ready.')
```

```python
# CELL 20: SAVE TO GOOGLE DRIVE  -- run this before disconnecting!
import shutil, os
from pathlib import Path

COLAB_PROJECT = '/content/pv_fault_detection'
DRIVE_SAVE    = '/content/drive/MyDrive/pv_fault_detection'
os.makedirs(DRIVE_SAVE, exist_ok=True)

def sync(src, label):
    s = Path(COLAB_PROJECT) / src
    d = Path(DRIVE_SAVE) / src
    if not s.exists():
        print(f'  SKIP {label} (not found)')
        return
    if d.exists():
        shutil.rmtree(d)
    shutil.copytree(s, d)
    mb = sum(f.stat().st_size for f in d.rglob('*') if f.is_file()) / 1e6
    print(f'  SAVED {label}: {mb:.1f} MB -> {d}')

print('Saving to Drive...')
sync('results',     'results (metrics.json)')
sync('checkpoints', 'checkpoints (model weights)')
sync('reports',     'reports (tables)')
sync('experiments', 'experiments (registry)')
# Also save the index so future sessions skip IRSM re-download
sync('data/processed', 'data/processed (index)')

print(f'\nSaved to: {DRIVE_SAVE}')
print('\nTo restore on your PC:')
print('  1. Download results/, checkpoints/, reports/ from Drive')
print('  2. Paste into C:\\BTP\\pv_fault_detection\\')
print('  3. Run this to update your local STATUS + tables:')
print('     python -m src.evaluation.report_tables --results_dir results/ --out reports/')
```
